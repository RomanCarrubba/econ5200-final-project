```python

import pandas as pd
from pathlib import Path

#settings and file pathes 

DATA_DIR = Path("data")
YEARS = list(range(2016, 2025))

PRICE_FILE = DATA_DIR / "pr_ex_mg.xlsx"
INDICATOR_FILE = DATA_DIR / "indicator_other.xlsx"
EV_FILE = DATA_DIR / "indicator_ev (1).xlsx"

# Make sure the AFDC data has been loaded.
if "laws" not in globals():
    raise NameError(
        "Load the AFDC laws and incentives CSV into a DataFrame named 'laws' first."
    )

# build the state year panel 

# Get the 50 states from the EV workbook.
state_table = pd.read_excel(
    EV_FILE,
    sheet_name="EV share",
    header=2
)

states = sorted(
    set(state_table["State"].dropna()) - {"DC", "US"}
)

if len(states) != 50:
    raise ValueError(
        f"Expected 50 states; found {len(states)}."
    )


def read_state_year_sheet(path, sheet_name, value_name):
    """Convert a state-by-year spreadsheet into a state-year dataframe."""

    wide = pd.read_excel(
        path,
        sheet_name=sheet_name,
        header=2
    )

    missing_years = set(YEARS) - set(wide.columns)

    if missing_years:
        raise ValueError(
            f"{sheet_name} is missing years: {sorted(missing_years)}"
        )

    wide = wide.loc[
        wide["State"].isin(states),
        ["State", *YEARS]
    ]

    long = wide.melt(
        id_vars="State",
        var_name="Year",
        value_name=value_name
    )

    long["Year"] = long["Year"].astype(int)

    return long


# Start with every state-year combination.
df = pd.DataFrame(
    [(state, year) for state in states for year in YEARS],
    columns=["State", "Year"]
)

# Add gasoline prices, population, and real GDP.
sheets = [
    (PRICE_FILE, "Prices", "Gasoline_Price"),
    (INDICATOR_FILE, "Total population", "Population_Thousands"),
    (INDICATOR_FILE, "Real GDP", "Real_GDP_Millions_2017"),
]

for path, sheet_name, value_name in sheets:
    values = read_state_year_sheet(
        path,
        sheet_name,
        value_name
    )

    df = df.merge(
        values,
        on=["State", "Year"],
        how="left",
        validate="one_to_one"
    )

# Add EV adoption share.
# The workbook reports this in percent, not as a 0-to-1 proportion.
ev_share = read_state_year_sheet(
    EV_FILE,
    "EV share",
    "EV_Share_Percent"
)

df = df.merge(
    ev_share,
    on=["State", "Year"],
    how="left",
    validate="one_to_one"
)

# Calculate real GDP per capita in dollars per person.
df["Real_GDP_Per_Capita"] = (
    df["Real_GDP_Millions_2017"]
    / df["Population_Thousands"]
    * 1000
)

df = df.drop(columns="Real_GDP_Millions_2017")


# build the ev purchase incentive treatment variable

policy_records = [
    {
        "State": "CT",
        "Title": "Electric Vehicle (EV) Rebate",
        "Start_Date": "2019-06-26",
        "End_Date": None,
    },
    {
        "State": "RI",
        "Title": "Zero Emission Vehicle (ZEV) and PHEV Rebates",
        "Start_Date": "2015-01-29",
        "End_Date": None,
    },
    {
        "State": "NY",
        "Title": "Electric Vehicle Rebate Program",
        "Start_Date": "2016-04-12",
        "End_Date": None,
    },
    {
        "State": "CO",
        "Title": "EV and Hydrogen Fuel Cell EV Tax Credit",
        "Start_Date": "2016-06-06",
        "End_Date": None,
    },
    {
        "State": "TX",
        "Title": "Light-Duty AFV Rebates",
        "Start_Date": "2017-06-12",
        "End_Date": None,
    },
    {
        "State": "OR",
        "Title": "PHEV and ZEV Rebates",
        "Start_Date": "2017-08-18",
        "End_Date": None,
    },
    {
        "State": "IL",
        "Title": "EV Rebates",
        "Start_Date": "2021-09-15",
        "End_Date": None,
    },
    {
        "State": "DE",
        "Title": "EV/PHEV Rebates",
        "Start_Date": "2023-08-03",
        "End_Date": None,
    },
    {
        "State": "WA",
        "Title": "Low-Income EV Rebates",
        "Start_Date": "2024-10-01",
        "End_Date": None,
    },
    {
        "State": "MD",
        "Title": "EV and Fuel Cell EV Tax Credit",
        # The program description indicates eligibility began
        # July 1, 2023, rather than on the 2010 enactment date.
        "Start_Date": "2023-07-01",
        "End_Date": None,
    },
]

policies = pd.DataFrame(policy_records)

policies["Start_Date"] = pd.to_datetime(
    policies["Start_Date"],
    errors="raise"
)

policies["End_Date"] = pd.to_datetime(
    policies["End_Date"],
    errors="coerce"
)

# Confirm that each manually selected policy appears in the AFDC data.
missing_states = set(policies["State"]) - set(laws["State"].dropna())

if missing_states:
    raise ValueError(
        f"These policy states were not found in the AFDC data: {missing_states}"
    )

print("All policy states found in AFDC data.")

# Create a treatment record for each year in which a policy overlaps
# at least part of that calendar year.
incentive_rows = []

for _, policy in policies.iterrows():

    for year in YEARS:

        year_start = pd.Timestamp(year=year, month=1, day=1)
        next_year = pd.Timestamp(year=year + 1, month=1, day=1)

        # A policy is active during a year if:
        # 1. It started before the following January 1.
        # 2. It has no recorded end date, or it ended on/after
        #    January 1 of the current year.
        active = (
            policy["Start_Date"] < next_year
            and (
                pd.isna(policy["End_Date"])
                or policy["End_Date"] >= year_start
            )
        )

        if active:
            incentive_rows.append(
                (policy["State"], year)
            )

# Convert to a state-year treatment dataframe.
incentives = pd.DataFrame(
    incentive_rows,
    columns=["State", "Year"]
).drop_duplicates()

incentives["EV_Incentive"] = 1

# Merge treatment into the main panel.
df = df.drop(
    columns="EV_Incentive",
    errors="ignore"
)

df = df.merge(
    incentives,
    on=["State", "Year"],
    how="left",
    validate="one_to_one"
)

# States without an active qualifying incentive receive zero.
df["EV_Incentive"] = (
    df["EV_Incentive"]
    .fillna(0)
    .astype("int8")
)



df.head()
```

```python
def balance_table(data, group, cols):
    """Means by group, the difference, and the standardised difference (difference / pooled SD).

    group: a 0/1 column. For RD pass the rows inside your bandwidth, grouped by side of the cutoff;
    for DiD, the pre-period rows grouped by treated state; for IV, grouped by the instrument.
    A standardised difference above about 0.1 in absolute value is worth a sentence in the proposal.
    """
    g1, g0 = data[data[group] == 1], data[data[group] == 0]
    rows = []
    for c in cols:
        sd = np.sqrt((g1[c].var() + g0[c].var()) / 2)
        rows.append({"variable": c, "treated": g1[c].mean(), "comparison": g0[c].mean(),
                     "difference": g1[c].mean() - g0[c].mean(),
                     "std. difference": (g1[c].mean() - g0[c].mean()) / sd if sd > 0 else np.nan})
    out = pd.DataFrame(rows).set_index("variable")
    out.loc["N"] = [len(g1), len(g0), np.nan, np.nan]
    return out.round(3)


# Whether a state ever had a qualifying EV incentive
df["treated"] = df.groupby("State")["EV_Incentive"].transform("max")
balance_table(df, "treated", ["EV_Share_Percent", "Gasoline_Price"])
```

### 3e. Sensitivity bound

How strong would an unobserved confounder have to be to erase the estimate, measured against one you observed?
For a regression-based estimate (DiD, DML's final stage, OLS with controls) the **robustness value** answers it
(Cinelli and Hazlett 2020, *JRSS-B* 82(1)). It is the share of the remaining variance of both the treatment and the
outcome that a confounder would have to explain to bring the estimate to zero. Compare it with the strongest control
you have: if no confounder as strong as that control could erase the estimate, say so; if one could, say that.

For RD and IV the equivalent question is about the design's own assumption: how far would the outcome have to jump
at the cutoff for other reasons, or how large a direct effect of the instrument, to explain the estimate?

```python
def robustness_value(t_stat, dof, q=1.0):
    """Cinelli and Hazlett's RV_q: the partial R-squared a confounder needs with BOTH treatment and outcome to
    move the estimate by 100*q percent (q = 1 brings it to zero)."""
    f = q * abs(t_stat) / np.sqrt(dof)
    return 0.5 * (np.sqrt(f**4 + 4 * f**2) - f**2)


def partial_r2(t_stat, dof):
    """A regressor's partial R-squared, recovered from its t-statistic."""
    return t_stat**2 / (t_stat**2 + dof)


# The paper's own example (Darfur, t = 4.18 on 783 degrees of freedom) gives RV = 13.9%:
print(f"check against Cinelli and Hazlett's example: RV = {robustness_value(4.18, 783):.1%} (paper: 13.9%)")

# Your estimate, from a regression with your controls (fit.tvalues / fit.df_resid):
# rv = robustness_value(fit.tvalues["treatment"], fit.df_resid)
# Benchmark: your strongest control's partial R-squared with the outcome, and with the treatment.
# out_fit = smf.ols("outcome ~ treatment + control_1 + control_2", data=df).fit()
# trt_fit = smf.ols("treatment ~ control_1 + control_2", data=df).fit()
# bench_y = partial_r2(out_fit.tvalues["control_1"], out_fit.df_resid)
# bench_d = partial_r2(trt_fit.tvalues["control_1"], trt_fit.df_resid)
# print(f"RV = {rv:.1%}; control_1 explains {bench_y:.1%} of the outcome and {bench_d:.1%} of the treatment")
```
